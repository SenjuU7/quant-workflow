```python
import pandas as pd
import numpy as np

df = pd.read_csv('D:/Visual Studio Code/Python/quant-workflow/data/export.csv')

print(df.info())
```

```python
def clean_number(value):
    if pd.isna(value):
        return np.nan

    value = value.replace(",","")

    if value.startswith("(") and value.endswith(")"):
        value = "-" + value[1:-1]

    return float(value)


numeric_columns = [
    "RETAIL SALES",
    "RETAIL TRANSFERS",
    "WAREHOUSE SALES"
]

for column in numeric_columns:
    df[column] = df[column].apply(clean_number)
```

```python
print(df.dtypes)
print(df[numeric_columns].isna().sum())
print(df[numeric_columns].describe())
```

```python
# Mengisi NaN di kolom ITEM TYPE dengan teks baru
df["ITEM TYPE"] = df["ITEM TYPE"].fillna("ALCOHOL")


print(
    df[df["ITEM CODE"] == "347939"]
    [["ITEM CODE", "ITEM DESCRIPTION", "ITEM TYPE"]]
)
```
